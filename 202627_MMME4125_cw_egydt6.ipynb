### Engineering problem, current practice and suitability of ML (3 marks)

The engineering problem is automatically identifying grain boundaries in metal microscopy images. Measuring grain size and shape supports quality control because these features influence mechanical properties.

Current methods include manual image analysis and electron backscatter diffraction (EBSD). Manual analysis is time-consuming and can vary between operators, while EBSD requires specialist equipment, careful sample preparation and a vacuum.

The paper provided demonstrate that polarised light microscopy combined with machine learning can identify grains and deformation twins. ML can learn colour, texture and boundary patterns from labelled images, potentially reducing manual work. Its accuracy must be checked using independent images.

### Dataset, variables and data quality (5 marks)
The selected dataset, `dataset_grains_ml_paper2`, contains **1,148 training images and 164 validation images** of metallic microstructures. These include rotated versions of some images, so not all images represent independent examples. Training images are used to teach the model, while validation images are used to assess its performance during development. No separate test folder is supplied.

The dataset contains three main components:

| Component | Meaning and purpose |
|---|---|
| **Images** | Coloured microscopy pictures used as model inputs. Each pixel contains red, green and blue intensity values, whose spatial arrangement represents the microstructure. |
| **Labels** | Text files containing coordinates that trace the outlines of annotated regions. |
| **Masks** | Black-and-white images showing boundary lines in black and interiors in white. |

The labels and masks provide reference answers for training and evaluating the model. Matching filenames associate each microscopy image with its annotations. Image–annotation pairs should be checked for completeness and alignment, particularly where images have been rotated.

#### Likely measurement and data-quality problems

Image noise, uneven illumination, polishing or etching quality, and limited spatial resolution may obscure boundaries. Small or thin features may be missed, while scratches could resemble genuine microstructural features. Differences in magnification may also change apparent grain size in pixels, so physical size measurements require spatial calibration.

Annotations may contain inaccurate outlines, omitted regions or inconsistent decisions about ambiguous boundaries. They should therefore be treated as potentially imperfect reference data.

#### Potential data leakage

Potential data leakage could occur if crops or rotated versions of the same image appear in both training and validation sets. This could allow the model to recognise familiar microstructures, giving an overly optimistic estimate of performance on unseen images. To prevent this, all versions of the same original image should remain in the same subset.

#### Plausible physical relationships

In polarised light microscopy, differences in crystal orientation can cause neighbouring grains to appear different in colour or brightness. These differences could help the model identify grain boundaries, although colour alone may not distinguish every grain.

Grain boundaries also restrict dislocation movement, so smaller grains generally increase yield strength within the range where the Hall–Petch relationship applies. This explains the engineering value of measuring grain size, although the supplied dataset does not include strength measurements.

```python
# Plot 1: microscopy images and matching masks

%matplotlib inline

import cv2
import numpy as np
import matplotlib.pyplot as plt
import random
from pathlib import Path

# Locate the training folders.
dataset = Path("dataset_grains_ml_paper2")
image_folder = dataset / "images" / "train"
mask_folder = dataset / "masks" / "train"

# Find images with matching masks.
matching_files = []

for file in sorted(image_folder.glob("*.png")):
    if (mask_folder / file.name).is_file():
        matching_files.append(file.name)

# Select 30 pairs using a repeatable random selection.
selected_files = random.Random(42).sample(matching_files, 30)

# Load the images and masks.
images = []
masks = []

for filename in selected_files:
    image = cv2.imread(str(image_folder / filename))
    mask = cv2.imread(
        str(mask_folder / filename), cv2.IMREAD_GRAYSCALE
    )

    # Check that both files loaded.
    if image is None or mask is None:
        raise ValueError(f"Could not load: {filename}")

    images.append(image)
    masks.append(mask)

# Display ten pairs.
plt.figure(figsize=(10, 26))

for i in range(10):
    plt.subplot(10, 2, 2 * i + 1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(selected_files[i])
    plt.axis("off")

    plt.subplot(10, 2, 2 * i + 2)
    plt.imshow(masks[i], cmap="gray", vmin=0, vmax=255)
    plt.title("Matching boundary mask")
    plt.axis("off")

plt.suptitle("Plot 1: microscopy images and matching masks")
plt.tight_layout(rect=[0, 0, 1, 0.97])
plt.show()
```

```python
# Plot 2: distribution of average image brightness

# Convert all images to greyscale and calculate their average brightness.
mean_brightness = []

for file in sorted(image_folder.glob("*.png")):
    gray_image = cv2.imread(str(file), cv2.IMREAD_GRAYSCALE)

    if gray_image is None:
        print(f"Could not read: {file.name}")
        continue

    mean_brightness.append(np.mean(gray_image))

print(f"Training images analysed: {len(mean_brightness)}")

# Display the brightness distribution.
plt.figure(figsize=(8, 5))

plt.hist(
    mean_brightness,
    bins=100
,
    color="steelblue",
    edgecolor="black"
)

plt.title("Plot 2: Brightness distribution of all training images")
plt.xlabel("Mean brightness (0 = black, 255 = white)")
plt.ylabel("Number of images")
plt.tight_layout()
plt.show()
```

```python
# Insert your code here for plot 3

# Count black and white pixels across the training mask folder.
boundary_pixels = 0
interior_pixels = 0
masks_used = 0

for file in sorted(mask_folder.glob("*.png")):
    mask = cv2.imread(str(file), cv2.IMREAD_GRAYSCALE)

    if mask is None:
        print(f"Could not read mask: {file.name}")
        continue

    if not np.isin(mask, [0, 255]).all():
        print(f"Skipped non-binary mask: {file.name}")
        continue

    boundary_pixels += np.count_nonzero(mask == 0)
    interior_pixels += np.count_nonzero(mask == 255)
    masks_used += 1

if masks_used == 0:
    raise ValueError("No readable binary masks found in the training folder.")

print(f"Training masks analysed: {masks_used}")
# 2. Convert the counts into percentages.
total_pixels = boundary_pixels + interior_pixels

boundary_percentage = 100 * boundary_pixels / total_pixels
interior_percentage = 100 * interior_pixels / total_pixels

percentages = [boundary_percentage, interior_percentage]

# 3. Display the percentages.
plt.figure(figsize=(7, 5))

plt.bar(
    ["Boundary markings", "Interiors"],
    percentages,
    color=["dimgray", "lightsteelblue"],
    edgecolor="black"
)

for i, percentage in enumerate(percentages):
    plt.text(
        i,
        percentage + 2,
        f"{percentage:.1f}%",
        ha="center"
    )

plt.title("Plot 3: Pixel-class balance across training masks")
plt.ylabel("Percentage of mask pixels")
plt.ylim(0, 110)
plt.tight_layout()
plt.show()
```
